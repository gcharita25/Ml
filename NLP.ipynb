```python
#1st program
import nltk

# Download required NLTK packages
nltk.download('punkt')
nltk.download('punkt_tab')

# Input text
text = "Natural Language Processing is an important field of Artificial Intelligence and it is used in many applications."

# Word Tokenization
words = nltk.word_tokenize(text)
print("Word Tokenization:")
print(words)

# Sentence Tokenization
sentences = nltk.sent_tokenize(text)
print("\nSentence Tokenization:")
print(sentences)
```

```python
#2nd program
import nltk
from nltk.stem import PorterStemmer

# Create an object of PorterStemmer
ps = PorterStemmer()

# List of words
words = ["playing", "played", "plays", "studies",
         "studying", "connected", "connection", "happiness"]

# Apply Porter Stemming
print("Original Word\tStemmed Word")
print("--------------------------------")

for word in words:
    stem = ps.stem(word)
    print(word, "\t\t", stem)
```

```python
#3rd program
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download required NLTK packages
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Original sentence
text = "Natural Language Processing helps computers understand human languages. NLP is widely used in chatbots and machine translation."

# Tokenization
words = word_tokenize(text)

# Load English stop words
stop_words = set(stopwords.words('english'))

# Identify stop words
found_stopwords = [word for word in words if word.lower() in stop_words]

# Remove stop words
filtered_words = [word for word in words if word.lower() not in stop_words]

# Display results
print("Original Sentence:")
print(text)

print("\nTokenized Words:")
print(words)

print("\nStop Words:")
print(found_stopwords)

print("\nWords After Stop Word Removal:")
print(filtered_words)
```

```python
#4th program
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from collections import Counter
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# Download packages
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Given tweets
tweets = [
    "I love this new smartphone!",
    "The movie was very boring and disappointing.",
    "The weather is pleasant today.",
    "This product is amazing and useful."
]

# Combine and convert to lowercase
text = " ".join(tweets).lower()

# Tokenize the text
words = word_tokenize(text)

# Remove punctuation
words = [word for word in words if word.isalpha()]

# Stop word removal
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

# Word frequency
word_freq = Counter(filtered_words)

print("Original Tweets:", tweets)
print("\nTokenized Words:", words)
print("\nFiltered Words:", filtered_words)
print("\nWord Frequencies:", word_freq)

# Generate and display Word Cloud
wc = WordCloud(width=800, height=400,
               background_color='white').generate_from_frequencies(word_freq)

plt.imshow(wc)
plt.axis('off')
plt.show()
```

```python
#5th program
import nltk
import re
from nltk.tokenize import word_tokenize
from collections import Counter

# Download packages
nltk.download('punkt')
nltk.download('punkt_tab')

# Given tweets
tweets = [
    "@john I love Python programming! 123",
    "@alice NLP is an interesting subject.",
    "Python is easy to learn!!! #coding",
    "@sam I love machine learning and AI."
]

# Remove usernames, special characters and numbers
cleaned = [re.sub(r'[^A-Za-z\s]', '',
          re.sub(r'@\w+', '', t)).lower() for t in tweets]

# Tokenization
words = []
for t in cleaned:
    words.extend(word_tokenize(t))

# Count word frequencies
freq = Counter(words)

# Display results
print("Preprocessed Tweets:", cleaned)
print("\nTokenized Words:", words)
print("\nWord Frequencies:")

for word, count in freq.items():
    print(word, ":", count)
```
